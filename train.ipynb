# 🚀 Обучение финального ансамбля моделей — «Защищённый смартфон»

В этом ноутбуке реализован полный цикл обучения ансамбля градиентных бустингов (CatBoost + LightGBM + XGBoost) с валидацией Stratified 5-Fold CV и калибровкой порога отсечения.


```python
import pandas as pd
import numpy as np
from src.models import InsurancePropensityEnsemble
from src.utils import seed_everything

# 1. Фиксация seed для воспроизводимости
SEED = 42
seed_everything(SEED)

# 2. Загрузка обучающих данных
train_df = pd.read_csv('train.csv')
print(f"Размер обучающей выборки: {train_df.shape}")

# 3. Инициализация и обучение ансамбля
ensemble = InsurancePropensityEnsemble(n_splits=5, seed=SEED)
cv_metrics = ensemble.fit(train_df)

# 4. Сохранение весов и артефактов
ensemble.save('models')
print("Обучение завершено. Веса сохранены в директорию models/")
```

## Результаты валидации Out-Of-Fold


```python
print(f"OOF ROC-AUC  : {cv_metrics['ensemble']['roc_auc']:.5f}")
print(f"OOF F1-Score : {cv_metrics['ensemble']['f1_score']:.5f}")
print(f"Accuracy     : {cv_metrics['ensemble']['accuracy']:.5f}")
print(f"Precision    : {cv_metrics['ensemble']['precision']:.5f}")
print(f"Recall       : {cv_metrics['ensemble']['recall']:.5f}")
print(f"Калиброванный порог: {cv_metrics['optimal_threshold']:.4f}")
```
