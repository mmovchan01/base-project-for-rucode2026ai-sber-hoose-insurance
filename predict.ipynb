# 🎯 Инференс и формирование файла предсказаний (submission)

Ноутбук загружает сохраненные веса модели из `models/`, выполняет feature engineering и генерирует `submission_seed_42.csv` без переобучения моделей.


```python
import pandas as pd
from src.models import InsurancePropensityEnsemble

# 1. Загрузка обученного ансамбля
ensemble = InsurancePropensityEnsemble.load('models')
print(f"Модель загружена. Seed: {ensemble.seed}, Порог: {ensemble.optimal_threshold:.4f}")

# 2. Загрузка тестовой выборки
test_df = pd.read_csv('public_test.csv')
print(f"Размер тестовой выборки: {test_df.shape}")

# 3. Предсказание
probs = ensemble.predict_proba(test_df)
preds = (probs >= ensemble.optimal_threshold).astype(int)

# 4. Формирование submission
submission = pd.DataFrame({
    'customer_id': test_df['customer_id'],
    'accepted': preds
})

output_file = f"submission_seed_{ensemble.seed}.csv"
submission.to_csv(output_file, index=False)
print(f"Файл предсказаний успешно сохранен: {output_file}")
print("Распределение предсказанных классов:")
print(submission['accepted'].value_counts(normalize=True))
submission.head(10)
```
